## Introduction to Tensors

### Creating tensors

PyTorch tensors are created using torch.tensor() -- https://docs.pytorch.org/docs/stable/tensors.html

```python
# scalar
scalar = torch.tensor(7)
print(scalar)
print(scalar.ndim)
# Get scalar back an Python int -- only works with scalar tensors
print(scalar.item())
```

### Rabdom tensors

Why random tensors?

They are important, because the way many neural networks learn is that they start with tensors full of random numbers and then adjust those random numbers to better represent the data.

`Start with some numbers -> look at data -> update the numbers -> repeat`

```python
# Create a random tensor of shape (3,4)
random_tensor = torch.rand(3,4)

# As we can see below, the entries are drawn from Unif(0,1)
random_tensor
```

```python
# Create a random tensor representing an image
random_image_size_tensor = torch.rand(3, 224, 224) # RGB color channels, height, width
random_image_size_tensor
```

### Tensor datatypes

**Note:** One of the most frequent errors when using PyTorch are:
1. wrong datatype
2. wrong shape
3. tensors on wrong device

```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype = None, # what datatype is the tensor (e.g. float32, float16, complex32)
                               device = None, # what device is your tensor on, eg. cpu, cuda
                               requires_grad=False) # whether or not to track gradients with this tensors operations
float_32_tensor
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
(float_16_tensor * float_32_tensor).dtype
```

### Manipulating Tensors (tensor operations)

Tensor operations include:
* Addition (el-wise)
* Multiplication (el-wise)
* Matrix multiplication

```python
# Matrix multiplication
tensor = torch.arange(1000)
```

```python
%%time
torch.matmul(tensor, tensor)
```

```python
%%time
value = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]
print(value)
```

```python
x = torch.tensor([[5,7,1],
                [100,200,5]])
print(x.max(), x.argmax()) # as you can see, argmax works weird with tensors of ndim>1,
                           # since it treats it as if it was of ndim=1
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - returns a view (**'pointer'**) of an input tensor of certain shape but **shares the memory with the original tensor**
* Stacking - combines multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - removes all 1-dimensions from a tensor
* Unsqueeze - adds a 1-dimension to a target tensor
* Permute - returns a view of the input with dimension permuted in a certain way

```python
x = torch.arange(1,10)
x, x.shape
```

```python
# Reshape
x_reshaped = x.reshape(3,3)
x_reshaped
```

```python
 # Change the view
 z = x.view(3,3)
 z[1:3,0] = 0
 x
```

```python
x_stacked = torch.stack([x,x,x,x], dim=1)
x_stacked
```

## PyTorch and NumPy

* Data in NumPy, want in PyTorch tensor -> torch.from_numpy(ndarray)
* PyTorch tensor, want in NumPy -> torch.your_tensor.numpy()

```python
import torch
import numpy as np

#numpy to tensor
array = np.arange(1.0,8.0)
tensor = torch.from_numpy(array)
array, tensor, array.dtype, tensor.dtype

# Notice that numpy array default data type is float64, whereas for torch tensor is float32.
# So if you want to get float32 you have to manually change the type
```

```python
# tensor to numpy

tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor, numpy_tensor.dtype
```

### Putting tensors (and models) on the GPU

```python
# Create a tensor (default on the CPU)
tensor = torch.tensor([1,2,3])
tensor, tensor.device
```

```python
# Move tensor to GPU (if available)
device = "cuda" if torch.cuda.is_available() else "cpu"
tensor_on_gpu = tensor.to(device)
tensor_on_gpu

# The above created a copy of tensor on GPU -- .device attribute if not modifiable
```

### NumPy only works with the CPU, so you might want to move tensors back to the CPU

```python
tensor_on_gpu.numpy()
```

```python
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
```
